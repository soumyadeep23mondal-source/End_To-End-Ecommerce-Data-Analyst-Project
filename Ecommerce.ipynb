```python
import pandas as pd
import numpy as np
```

```python
# remove duplicates
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df
```

```python
df['Discount'] = pd.to_numeric(df['Discount'], errors='coerce')
```

```python
# remove extra space
df = df.apply(lambda x: x.str.strip() if x.dtype == "object" else x)
```

```python
df
```

```python
# standardize text format
# convart text into proper case
df['Customer_Name'] = df['Customer_Name'].str.title()

df['City'] = df['City'].str.title()

df['State'] = df['State'].str.title()
```

```python
# validate email addresses
# keep only valide email record

df = df[df['Email'].str.contains('@',na=False)]
```

```python
df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce',dayfirst=True)
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce',dayfirst=True)
```

```python
# Convert other numeric columns
df['Qty'] = pd.to_numeric(df['Qty'], errors='coerce')
df['Unit_Price'] = pd.to_numeric(df['Unit_Price'], errors='coerce')
```

```python
# remove invalid quantity
# remove records having negative or zero quantity

df = df[df['Qty']>0]
```

```python
df['Discount'] = df['Discount'].fillna(0)

df['Phone'].fillna('Unknown')

df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
# Feature engineering
# total Sales

df['Sales'] = df['Qty'] * df['Unit_Price']

```

```python
# net Account
df['Net_Account'] = df['Sales'] - (df['Sales']*df['Discount']/100)
```

```python
# profite (example:20)
df['Profit'] = df['Net_Account']*0.20
```

```python
# Order MOnth
df['Order_Month'] = df['Order_Date'].dt.month_name()
```

```python
# order year
df['Order_Year'] = df['Order_Date'].dt.year
```

```python
# weekday
df['Weekday'] = df['Order_Date'].dt.day_name()
```
